## Finding common IDs between the two datasets

```python
common_ids = set(monzo_calcs['id']).intersection(monzoapi_calcs['id'])
len(common_ids)
```

## Finding common columns between the two datasets

```python
common_cols = list(set(monzo_calcs.columns).intersection(set(monzoapi_calcs.columns)))
common_cols
```

## Filtering out column with no interest for this task

Ideally we want to keep only leaf rules, that act upon the basic transaction fields

```python
fields = ['amount', 'datetime', 'description', 'amount_cur', 'currency', 'tags']

cols_of_interest = common_cols.copy()

for col in common_cols:
    for field in fields:
        if col == field:
            print(f"REMOVED: column '{col}' is field")
            cols_of_interest.remove(col)
            break
        elif col.startswith(field+'.'):
            print(f"REMOVED: column '{col}' is transformation")
            cols_of_interest.remove(col)
            break
        elif col.startswith('tags_list') or col.startswith('tags_added') or col.startswith('new_tags_list'):
            print(f"REMOVED: column '{col}' is tags column")
            cols_of_interest.remove(col)
            break
        elif '_tags' in col:
            print(f"REMOVED: column '{col}' is _tags column")
            cols_of_interest.remove(col)
            break
        # else:
        #    print(f"ADDED: column {col} is VALID")
cols_of_interest
```

## creating the filtered datasets

```python
monzo = monzo_calcs[monzo_calcs['id'].isin(common_ids)].sort_values('id')[cols_of_interest].reset_index(drop=True)
monzoapi = monzoapi_calcs[monzoapi_calcs['id'].isin(common_ids)].sort_values('id')[cols_of_interest].reset_index(drop=True)
monzo.shape, monzoapi.shape
```

## Comparison dataframe

```python
comps = monzo.compare(monzoapi)
# comps.columns = [c[0]+' ('+('monzo' if c[1]=='self' else 'monzoapi')+')' for c in comps.columns]
# cols = list({c for c in comps.columns})
cols = list({c[0] for c in comps.columns})

comps['id'] = monzo['id']
comps
```

```python
# cols = comps.columns
# cols = list({c[0] for c in comps.columns})
cols
```

## get_col function to fetch the old and the new version of a given leaf rule

```python
def get_col(col):
    res = comps[~comps[col].isna().any(axis=1)][['id', col]]
    res.columns = ['id', 'monzo', 'monzoapi']
    res['column'] = col
    res['count'] = len(res)
    return res

get_col(cols[1])
```

## stack all diffs together in a tidy format

```python
all_diffs = pd.concat([get_col(col) for col in set(cols).difference(['id'])]).sort_values('count', ascending=False)
all_diffs
```

## Columns with the most discrepancy

```python
all_diffs.groupby(['column']).agg({'count':'mean'}).reset_index().sort_values('count', ascending=False)
```
